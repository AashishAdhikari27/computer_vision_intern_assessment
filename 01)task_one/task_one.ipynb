```python
# importing libraries 

import pandas as pd
import numpy as np

import cv2 ##  importing opencv lirary 
# we can just import it as "import cv2 as cv " only too 

import matplotlib.pyplot as plt
```

```python

image_path = '01)task_one/bear.jpg'

# loading image using opencv
image = cv2.imread(image_path)
```

```python

print(image)          # prints the raw pixel values (as a NumPy array)

print(type(image))    # confirms the data type 
```

```python

# to visualize the image in a separate popup window, which is preferred for .py files 
cv2.imshow('Bear Image', image)
cv2.waitKey(0)      # Waits for any key press
cv2.destroyAllWindows()  # Closes all windows
```

```python
#for the jupyter notebook (i.e for .ipynb files) we can view the image in the notebook using matplotlib 

#As imread() method loads the image in the BGR 8-bit format which is default.So, converting BGR to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.title('Bear Image')
plt.axis('off')
plt.show()
```

# Converting image in grayscale

```python

gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)  # convers the bgr image to gray image using .cvtColor() function 
# for documentation : https://docs.opencv.org/4.11.0/d8/d01/group__imgproc__color__conversions.html#gga4e0972be5de079fed4e3a10e24ef5ef0a353a4b8db9040165db4dacb5bcefb6ea

cv2.imwrite('01)task_one/output_images/bear_gray.jpg', gray_image)  # this saves the converted gray image to desired file location with desired name 
```

```python

# Display the grayscale image using matplotlib
plt.imshow(gray_image, cmap='gray')  # 'cmap=gray' is important for grayscale images
plt.title('Grayscale Image')         # Optional title
plt.axis('off')                      # Hides the axes for better visual
plt.show()
```

# Applying gaussian blur
In Gaussian Blur, instead of using a simple average (box filter), we use a Gaussian kernel, which gives more weight to the center pixels. This is done using the function cv2.GaussianBlur().

    We need to give the width and height of the kernel, and both must be positive odd numbers (like 3, 5, 7, etc.).

    We also set values for sigmaX and sigmaY, which control how much the image is blurred in the horizontal and vertical directions.

    If you only set sigmaX, OpenCV uses the same value for sigmaY.

    If you set both as 0, OpenCV calculates them automatically based on the kernel size.

Gaussian blur is especially good at removing noise that follows a Gaussian distribution (common in real-world images).

--> view this documentation for more details : https://docs.opencv.org/4.x/d4/d13/tutorial_py_filtering.html

```python
# Apply Gaussian Blur to the grayscale image
blurred_image = cv2.GaussianBlur(gray_image, (5, 5), 0)  # kernel size = (5,5), sigmaX = 0

# Save the blurred image
cv2.imwrite('01)task_one/output_images/bear_blur.jpg', blurred_image)
```

```python

# Display the blurred gray image using matplotlib
plt.imshow(blurred_image, cmap='gray')  # 'cmap=gray' ensures correct display for grayscale images
plt.title('Gaussian Blurred Image')
plt.axis('off')  # Hides the axis for a cleaner look
plt.show()
```
